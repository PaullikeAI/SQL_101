## 11. Views and Subqueries: CREATE VIEW, DROP VIEW

In this notebook creating views and subqueries and their uses will be covered.

###### Keywords Covered - CREATE VIEW, DROP VIEW

```python
import sqlite3
import pandas as pd

# Connect to database
data_conn = sqlite3.connect("pokemon.db")
```

### CREATE VIEW

Use: Creating an object out of a query to make it reuseable. This ensures consistency when doing the same query multiple times. It also enhances security by restriciting access to specific data in the table/tables.

*Syntax: CREATE VIEW view_name AS*

```python
# Create a view which selects just the fire Pokemon.
data_conn.execute(""" CREATE VIEW fire_pokemon AS
                      SELECT * 
                      FROM pokemon
                      WHERE type1 = 'fire'""")
```

```python
# Display the data in the view.
pd.read_sql("""SELECT * 
               FROM fire_pokemon
               LIMIT 12""", con=data_conn)
```

### DROP VIEW

Use: Deletes the view.

```python
# Delete the fire_pokemon view.
data_conn.execute("""DROP VIEW fire_pokemon""")
```

```python
# Try to display the data in the view.
pd.read_sql("""SELECT * 
               FROM fire_pokemon
               LIMIT 12""", con=data_conn)
```

This now raises an error as the view has been dropped.

### Subqueries

Use: A subquery is a query nested within another query. These can have uses like:

*    Applying aggregate functions (e.g AVG, SUM) dynamically.
*    Selecting rows based on the requirements returned by the subquery.

### Single row Subqueries

These are subqueries which return one result.

Below we will get the largest height value, and then use that to return the data on the pokemon with that height.

```python
# View the data on the tallest Pokemon.
pd.read_sql("""SELECT id, name,	type1, type2, height, weight, dex_entry
               FROM pokemon
               WHERE height = (SELECT MAX(height) from pokemon)""", con=data_conn)
```

### Multi Row Subqueries

Use: A subquery that returns many rows based on the double query.

Below we will return all fire Pokemon in the first 250 Pokemon (main query) that are above average weight (subquery).

```python
pd.read_sql("""SELECT id, name,	type1, height, weight
               FROM pokemon
               WHERE weight > (SELECT AVG(weight) from pokemon)
               AND id <= 250
               AND type1 = 'fire'""", con=data_conn)
```
